# Translation and review software check

This notebook translates one synthetic triplet with a local provider, binds fixture review decisions to exact text, and builds a contract-validated release. The release remains pending human qualification.

Install the paired Fair-UK checkout and this repository with the documented optional dependencies before running. No network requests or checkpoint downloads occur.


```python
import tempfile
from pathlib import Path

from fairforget.examples import run_translation_example

work = Path(tempfile.mkdtemp(prefix="fairforget-software-check-"))
prompt_root = next(
    base / "prompts"
    for base in (Path.cwd(), *Path.cwd().parents)
    if (base / "prompts/manifest.json").is_file()
)
```

```python
result = run_translation_example(work / "run", prompt_root=prompt_root)
assert result["qualification"] == "synthetic_software_check"
result
```

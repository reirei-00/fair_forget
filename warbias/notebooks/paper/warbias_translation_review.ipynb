# Ukrainian-to-English translation

Translate existing Ukrainian triplets into English. Keep both versions with matching item IDs. Review the translations, then build a paired release.

This offline example uses synthetic inputs, responses and ratings. See the [install guide](../../README.md) and [pipeline commands](../../docs/warbias-construction.md).


```python
import tempfile
from pathlib import Path

from fairforget.examples import run_translation_example

work = Path(tempfile.mkdtemp(prefix="fairforget-software-check-"))
prompt_root = next(
    candidate / "prompts"
    for base in (Path.cwd(), *Path.cwd().parents)
    for candidate in (base / "warbias", base)
    if (candidate / "prompts/manifest.json").is_file()
)
```

```python
result = run_translation_example(work / "run", prompt_root=prompt_root)
assert result["qualification"] == "synthetic_software_check"
result
```
